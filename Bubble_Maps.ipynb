1.1 Scatter, Bbble Correlation

​Relationship between two numerical variables. ​

```python
import pandas as pd
import numpy as np
import plotly.express as px

np.random.seed(42)

watch_time = np.random.randint(2, 25, 100)

scores = watch_time * 3 + np.random.normal(20, 8, 100)

df = pd.DataFrame({
    "Watch_Time": watch_time,
    "Score": scores
})
```

Suppose you are analyzing an online learning platform. ​You want to know does spending more time watching lectures improve assessment scores? ​Do learners who practice more perform better? ​

```python
fig = px.scatter(
    df,
    x="Watch_Time",
    y="Score",
    title="Watch Time vs. Assessment Score"
)

fig.show()
```

see slowly all the data points generally move upwards, going upwards. ​This is a kind of trend. ​And it suggests learners who spend more time watching videos tend to score higher. ​This is the meaning of that. ​This is called positive relationship.

```python
fig = px.scatter(
    df,
    x="Watch_Time",
    y="Score",
    trendline="ols",
    title="Watch Time vs. Assessment Score"
)

fig.show()
```

```python
df["Assignments"] = np.random.randint(1, 15, 100)
```

```python
fig = px.scatter(
    df,
    x="Watch_Time",
    y="Score",
    size="Assignments",
    color="Assignments",
    title="Learner Performance"
)

fig.show()
```

1.3 Heatmap & Density Visuals

imagine you are managing an online learning platform you want to ​know when our student most active which day has the highest engagement and you ​know what time should live class be scheduled

```python
import pandas as pd
import plotly.express as px
import numpy as np

np.random.seed(42)

days = ["Mon","Tue","Wed","Thu","Fri","Sat","Sun"]
hours = list(range(24))

data = []

for day in days:
    for hour in hours:
        data.append({
            "Day": day,
            "Hour": hour,
            "Activity": np.random.randint(20,150)
        })

df = pd.DataFrame(data)
```

```python
fig = px.density_heatmap(
    df,
    x="Hour",
    y="Day",
    z="Activity",
    color_continuous_scale="Viridis",
    title="Learner Activity by Hour and Day"
)

fig.show()
```

by the density of the ​color you can understand the things the dark color means lower count lower ​activity and the yellow color or light color means the higher activity /  a heat map chart each cell tells us how active learners ​were during a specific hour on a particular day reading a heat map guys ​suppose the darker cells appear around 8 p.m. 9 p.m. 10 p.m

we cannot conclude that ​students study more on weekends this information can help us schedule live ​sessions send notification at optimal times release new courses when ​engagement is highs and a heat map transforms thousands of data points into ​a simple visual story

```python
df2 = pd.DataFrame({
    "Watch_Time": np.random.normal(12, 3, 1000),
    "Score": np.random.normal(75, 10, 1000)
})
```

```python
fig = px.density_heatmap(
    df2,
    x="Watch_Time",
    y="Score",
    title="Watch Time vs Assessment Score"
)

fig.show()
```

1.4 Choropleth maps

​A choropleth map colors different geographical regions based on a numerical value. ​For example, number of learners by country, revenue by state, population by region, sales ​by district.

```python
import pandas as pd
import plotly.express as px

df = pd.DataFrame({
    "Country": [
        "India",
        "United States",
        "Canada",
        "Australia",
        "Germany"
    ],

    "ISO": [
        "IND",
        "USA",
        "CAN",
        "AUS",
        "DEU"
    ],

    "Learners": [
        5200,
        3400,
        1800,
        1500,
        2200
    ]
})
```

```python
fig = px.choropleth(
    df,
    locations="ISO",
    color="Learners",
    hover_name="Country",
    color_continuous_scale="Blues",
    title="Learners by Country"
)

fig.show()
```

It is showing the world map and in world map it is showing the color scale for different ​number of values guys. ​So notice that we have included both country names and ISO country code.

```python
fig = px.choropleth(
    df,
    locations="ISO",
    color="Learners",
    hover_name="Country",
    hover_data={
        "Learners": True
    }
)
fig = px.choropleth(
    df,
    locations="ISO",
    color="Learners",
    hover_name="Country",
    hover_data={
        "Learners": True
    }
)
```

```python
fig.show()
```

In color scale we have yellow and darker side we have blue. ​In between we have a scale of yellow and red. ​Shade is there. ​So when we are showing the yellow color that means in India we have minimum learners are ​there.

​Large countries like Canada, Australia occupy more space on the map. ​This does not mean the necessary have more learners.
​Always interpret the color not the physical area of the country okay. ​Always remember that point. ​

1.5 Scatter Bubble Maps for Learner Hubs

A choropleth map shows values by state or country. But if you want to show the exact locations of learners, a scatter map is better. Each location is shown as a point, such as a city, training center, or learner hub. This gives more precise geographical information.

```python
import pandas as pd

df = pd.DataFrame({

    "City": [
        "Ahmedabad",
        "Mumbai",
        "Delhi",
        "Bangalore",
        "Hyderabad"
    ],

    "Latitude": [
        23.0225,
        19.0760,
        28.6139,
        12.9716,
        17.3850
    ],

   "Longitude": [
        72.5714,
        72.8777,
        77.2090,
        77.5946,
        78.4867
    ],

    "Enrollments": [
        1200,
        2500,
        1800,
        2200,
        1500

    ]

})
```

```python
px.scatter_geo()
```

```python
import plotly.express as px

fig = px.scatter_geo(
    df,
    lat="Latitude",
    lon="Longitude",
    hover_name="City",
    size="Enrollments",
    title="Learner Hubs Across India"
)

fig.show()
```

Each city appears as a bubble. The bubble ​size represents number of enrollments. Larger bubbles indicates more learners. Small bubbles ​indicates fewer learners, okay. Immediately, we can identify major learners and hubs also.

```python
fig = px.scatter_geo(
    df,
    lat="Latitude",
    lon="Longitude",
    size="Enrollments",
    color="Enrollments",
    hover_name="City",
    hover_data={
        "Enrollments": True
    }
)
```

```python
import plotly.express as px

fig = px.scatter_geo(
df,
    lat="Latitude",
    lon="Longitude",
    size="Enrollments",
    color="Enrollments",
    hover_name="City",
    height=600
)

fig.show()
```

1.6 Density Maps Region

Q: We ask where are learners concentrated,

```python
import pandas as pd
import numpy as np
import plotly.express as px

np.random.seed(42)

df = pd.DataFrame({

    "Latitude": np.random.uniform(8, 37, 1000),

    "Longitude": np.random.uniform(68, 97, 1000),

    "Enrollments": np.random.randint(1, 50, 1000)

})
```

```python
import plotly.express as px
import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({

    "Latitude": np.random.uniform(8, 37, 1000),

    "Longitude": np.random.uniform(68, 97, 1000),

    "Enrollments": np.random.randint(1, 50, 1000)

})

fig = px.density_map(
    df,
    lat="Latitude",
    lon="Longitude",
    z="Enrollments",
    radius=20,
    zoom=4,
    map_style="open-street-map",
    title="Learner Density Across India"
)

fig.show()
```

1.7 Animated interactive Maps

Q: ​How did our learner network grow over the year? ​

```python
import pandas as pd

df = pd.DataFrame({

    "Year": [
        2021, 2021, 2021,
        2022, 2022, 2022,
        2023, 2023, 2023,
        2024, 2024, 2024
    ],

    "City": [
        "Ahmedabad", "Mumbai", "Delhi"
    ] * 4,

    "Latitude": [
        23.0225, 19.0760, 28.6139
    ] * 4,

   "Longitude": [
        72.5714, 72.8777, 77.2090
    ] * 4,

    "Enrollments": [
        500, 700, 400,
        900, 1200, 800,
        1500, 1900, 1400,
        2300, 2800, 2200
    ]
})
```

```python
import plotly.express as px

fig = px.scatter_map(
    df,
    lat="Latitude",
    lon="Longitude",
    size="Enrollments",
    color="Enrollments",
    hover_name="City",
    animation_frame="Year",
    zoom=4,
    map_style="open-street-map",
    title="Learner Growth Across India"
)

fig.show()
```
