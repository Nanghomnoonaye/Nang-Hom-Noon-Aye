```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

x = np.arange(1, 11)
low_disp = np.random.normal(loc=0, scale=0.5, size=10)
high_disp = np.random.normal(loc=0, scale=3.0, size=10)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Symmetric y-limits force line 0 directly to vertical center
y_max = max(np.abs(high_disp).max(), np.abs(low_disp).max()) + 1

for ax, data in zip(axes, [low_disp, high_disp]):
    ax.plot(x, data, 'ko')
    ax.axhline(0, color='black', linewidth=1.5)
    ax.set_facecolor('#e8e8e8')
    ax.set_ylim(-y_max, y_max)
    ax.set_xticks([])
    ax.set_yticks([])

plt.suptitle("Descriptive statistics - Dispersion", fontsize=14, color='black')
plt.tight_layout()
plt.show()
```

```python
# The points are the data values or measurements.

# ​And the straight lines are your means.

You can see that the dispersion or ​the variation on the left is lower than the dispersion on the right.

how would we quantify this in a number?

```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 11)
low_disp = np.array([0.4, -0.6, 0.2, -0.3, 0.7, -0.2, 0.5, -0.3, 0.3, -0.5])
high_disp = np.array([2.5, 3.2, 0.2, -3.5, -3.8, 1.2, -2.1, -0.4, 1.1, 2.6])

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Symmetric limits to place line 0 exactly in the middle
y_max = max(np.abs(high_disp).max(), np.abs(low_disp).max()) + 0.5

for ax, data in zip(axes, [low_disp, high_disp]):
    # Red vertical deviation lines from mean line to points
    ax.vlines(x, 0, data, colors='#a81c3f', linewidth=2.5)
    # Dark blue center line
    ax.axhline(0, color='#0a1c54', linewidth=2)
    # Black dots
    ax.plot(x, data, 'ko', markersize=6)

    ax.set_facecolor('#e8e8e8')
    ax.set_ylim(-y_max, y_max)
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```

```python
We look for each point as the distance to the mean.

We usually average these distances to quantify dispersion.

# one Prob
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(1, 11)
low_disp = np.array([0.4, -0.6, 0.2, -0.3, 0.7, -0.2, 0.5, -0.3, 0.3, -0.5])
high_disp = np.array([2.5, 3.2, 0.2, -3.5, -3.8, 1.2, -2.1, -0.4, 1.1, 2.6])

fig, axes = plt.subplots(1, 2, figsize=(8, 3))
y_max = max(np.abs(high_disp).max(), np.abs(low_disp).max()) + 1.2

for ax, data in zip(axes, [low_disp, high_disp]):
    ax.axhline(0, color='#0a1c54', linewidth=2, zorder=2)
    ax.plot(x, data, 'ko', markersize=5, zorder=3)
    ax.set_ylim(-y_max, y_max)
    ax.axis('off')

# Left subplot annotations
axes[0].vlines(x[4], 0, low_disp[4], colors='#a81c3f', linewidth=2, zorder=1)
axes[0].vlines(x[9], 0, low_disp[9], colors='#a81c3f', linewidth=2, zorder=1)
axes[0].annotate('', xy=(x[4], low_disp[4]/2), xytext=(5.5, -3.8), arrowprops=dict(arrowstyle='->', color='#a81c3f', lw=1.5))
axes[0].annotate('', xy=(x[9], low_disp[9]/2), xytext=(5.5, -3.8), arrowprops=dict(arrowstyle='->', color='#a81c3f', lw=1.5))

# Right subplot annotations
axes[1].vlines(x[1], 0, high_disp[1], colors='#a81c3f', linewidth=2, zorder=1)
axes[1].vlines(x[6], 0, high_disp[6], colors='#a81c3f', linewidth=2, zorder=1)
axes[1].annotate('', xy=(x[1], high_disp[1]/2), xytext=(5.5, -3.8), arrowprops=dict(arrowstyle='->', color='#a81c3f', lw=1.5))
axes[1].annotate('', xy=(x[6], high_disp[6]/2), xytext=(5.5, -3.8), arrowprops=dict(arrowstyle='->', color='#a81c3f', lw=1.5))

# Subtitle labels
axes[0].text(5.5, -y_max, "Average zero", ha='center', fontsize=10)
axes[1].text(5.5, -y_max, "Average zero", ha='center', fontsize=10)

plt.tight_layout()
plt.show()
```

​We usually average these distances to quantify dispersion. ​There's one problem, however. ​As these deviations are both positive and ​negative, their average will be approximately zero.

 to solve this problem, ​we square the distance first because squaring will give us positive numbers. ​The average squared distance is called the **variance**, ​taking the square root again gives us the standard deviation. ​A **standard deviation** of 0.016, as in this example, ​can be interpreted as, that on average, ​observations are 0.016% away from the mean.

# M2-V5 Pareto

​80% of all problems can be attributed to 20% of the causes. ​Ever heard this saying before?, the 80/20 principle, also known as the Pareto ​principle, and also known as the rule of the vital few and the trivial many?

```python
#Example
```

Consider, you work in a **factory** that produces cookies called Cafe Noir. ​These are famous Dutch cookies and they are really, really good.

```python
# ​Sometimes the production line is stopped due to some problem.

Dta collet = Stop/ Duration / Causes
```

For each stop, you have recorded the duration of the stop in minutes ​as well as the cause for the stop.

ou start a project with the objective to **reduce the time lost ​due to production stops**. ​Therefore, you wish to select the cost that occurs most often. ​Using a Pareto chart, we can analyze this.

you go to the** Start > Quality Tools > Pareto Chart**. ​Now Minitab asks for you, what is your defects or ​attribute data, which is basically the category that you want to count. ​That is, causes. ​

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Data
df = pd.DataFrame({
    'Cause': [
        'dimensions', 'Break', 'dough', 'color',
        'recipe', '(small)', 'recipe', 'failure',
        'Experiment', 'No', 'belt', 'Other'
    ],
    'Count': [37, 27, 8, 4, 2, 2, 2, 1, 1, 1, 1, 4]
})

df['Percent'] = (df['Count'] / df['Count'].sum()) * 100
df['Cum_Percent'] = df['Percent'].cumsum()

# Plot setup
fig, ax1 = plt.subplots(figsize=(8, 4.5))

# Bars (Count)
ax1.bar(
    df['Cause'],
    df['Count'],
    color='#82A8D8',
    edgecolor='#444444',
    width=1.0,
    align='edge',
)
ax1.set_ylabel('Count', fontweight='bold', color='#444444')
ax1.set_ylim(0, 95)
ax1.grid(axis='y', color='#E5E5E5', lw=0.8)

# Line (Cumulative Percent)
ax2 = ax1.twinx()
x_pts = np.arange(len(df)) + 0.5
ax2.plot(
    x_pts,
    df['Cum_Percent'],
    color='#1F5CA9',
    marker='o',
    ms=4,
    lw=1,
)
ax2.set_ylabel('Percent', fontweight='bold', color='#444444')
ax2.set_ylim(0, 105)

# Styling & Labels
ax1.set_xticks(x_pts)
ax1.set_xticklabels(
    df['Cause'],
    rotation=45,
    ha='right',
    rotation_mode='anchor',
    fontweight='bold',
)
ax1.text(
    -0.12,
    -0.12,
    'Cause',
    transform=ax1.transAxes,
    fontweight='bold',
    color='#444444',
)
plt.title('Pareto Chart of Cause', fontsize=14, pad=10, color='#333333')

# Data table below chart
table_data = [
    [f'{x}' for x in df['Count']],
    [f'{x:.1f}' for x in df['Percent']],
    [f'{x:.1f}' for x in df['Cum_Percent']],
]
table = plt.table(
    cellText=table_data,
    rowLabels=['Count', 'Percent', 'Cum %'],
    loc='bottom',
    bbox=[0, -0.6, 1, 0.25],
    cellLoc='center',
    edges='',
)
table.auto_set_font_size(False)
table.set_fontsize(8)

plt.tight_layout()
plt.show()
```

cum % is the line above

​The curve shows the cumulative frequencies of the issues. ​These cumulative percentages can be read ​from the right vertical axis as well as from the bottom of the chart. ​From this chart, ​it can be concluded that the three largest issues cause **80% **of all problems. ​However, this Pareto chart with the **frequencies only shows ​how often each problem occurs. **
