```python
import matplotlib.patches as patches
import matplotlib.pyplot as plt

# Create Figure
fig, ax = plt.subplots(figsize=(10, 6.5), dpi=200)
ax.set_xlim(0, 10)
ax.set_ylim(0, 7)
ax.axis('off')

# Styling Colors
c_yellow = '#FFFFD1'
c_blue = '#CBEFF6'
c_pink = '#F5C2E7'
c_border = '#000000'


# Helper Function: Rectangle (Step / Operation)
def draw_rect(x, y, w, h, text):
  rect = patches.Rectangle(
      (x - w / 2, y - h / 2),
      w,
      h,
      facecolor=c_yellow,
      edgecolor=c_border,
      lw=1,
      zorder=2,
  )
  ax.add_patch(rect)
  ax.text(
      x,
      y,
      text,
      ha='center',
      va='center',
      fontsize=8,
      fontfamily='sans-serif',
      zorder=3,
  )


# Helper Function: Parallelogram (Input / Output)
def draw_para(x, y, w, h, text):
  slant = 0.12
  pts = [
      [x - w / 2 + slant, y + h / 2],
      [x + w / 2 + slant, y + h / 2],
      [x + w / 2 - slant, y - h / 2],
      [x - w / 2 - slant, y - h / 2],
  ]
  poly = patches.Polygon(
      pts, facecolor=c_blue, edgecolor=c_border, lw=1, zorder=2
  )
  ax.add_patch(poly)
  ax.text(
      x,
      y,
      text,
      ha='center',
      va='center',
      fontsize=8,
      fontweight='bold',
      zorder=3,
  )


# Helper Function: Diamond (Decision)
def draw_diamond(x, y, w, h, text=''):
  pts = [[x, y + h / 2], [x + w / 2, y], [x, y - h / 2], [x - w / 2, y]]
  poly = patches.Polygon(
      pts, facecolor=c_pink, edgecolor=c_border, lw=1, zorder=2
  )
  ax.add_patch(poly)
  if text:
    ax.text(
        x,
        y,
        text,
        ha='center',
        va='center',
        fontsize=8,
        fontweight='bold',
        zorder=3,
    )


# Helper Function: Straight Arrow
def draw_arrow(x1, y1, x2, y2):
  ax.annotate(
      '',
      xy=(x2, y2),
      xytext=(x1, y1),
      arrowprops=dict(
          arrowstyle='->', color='black', lw=1, mutation_scale=10
      ),
      zorder=1,
  )


# Helper Function: Orthogonal Loopback Arrow (Right -> Up -> Left)
def draw_loopback(x_start, y_start, x_corner, y_corner, x_end, y_end):
  ax.plot(
      [x_start, x_corner, x_corner, x_end],
      [y_start, y_start, y_corner, y_corner],
      color='black',
      lw=1,
      zorder=1,
  )
  ax.annotate(
      '',
      xy=(x_end, y_corner),
      xytext=(x_end + 0.15, y_corner),
      arrowprops=dict(
          arrowstyle='->', color='black', lw=1, mutation_scale=10
      ),
      zorder=1,
  )


# --- LEFT SIDE: Business Process ---
ax.text(
    2.0,
    6.4,
    'Business Process',
    fontsize=11,
    fontweight='bold',
    ha='center',
    va='center',
)

draw_para(0.9, 5.7, 0.8, 0.38, 'Input')
draw_rect(2.0, 5.7, 0.9, 0.38, 'Step')
draw_rect(2.0, 4.9, 0.9, 0.38, 'Step')
draw_diamond(2.0, 4.1, 0.9, 0.5, 'Decision')
draw_rect(2.0, 3.1, 0.9, 0.38, 'Step')
draw_rect(2.0, 2.3, 0.9, 0.38, 'Step')
draw_para(3.0, 2.0, 0.8, 0.38, 'Output')

# Left Straight Arrows
draw_arrow(1.3, 5.7, 1.55, 5.7)  # Input -> Step 1
draw_arrow(2.0, 5.51, 2.0, 5.09)  # Step 1 -> Step 2
draw_arrow(2.0, 4.71, 2.0, 4.35)  # Step 2 -> Decision
draw_arrow(2.0, 3.85, 2.0, 3.29)  # Decision -> Step 3
draw_arrow(2.0, 2.91, 2.0, 2.49)  # Step 3 -> Step 4
draw_arrow(2.45, 2.3, 2.6, 2.0)  # Step 4 -> Output

# Left Decision Loopback (Starts at right tip of diamond, goes up to Step 2)
draw_loopback(2.45, 4.1, 2.7, 4.9, 2.45, 4.9)


# --- RIGHT SIDE: Work Process Container ---
wp_box = patches.Rectangle(
    (4.7, 2.1),
    4.1,
    4.3,
    facecolor='none',
    edgecolor='black',
    lw=1,
    zorder=1,
)
ax.add_patch(wp_box)
ax.text(
    6.75,
    6.2,
    'Work Process',
    fontsize=11,
    fontweight='bold',
    ha='center',
    va='center',
)

draw_para(5.3, 5.7, 0.8, 0.38, 'Input')
draw_rect(6.4, 5.7, 0.9, 0.38, 'Operation/\nTask')
draw_rect(6.4, 4.9, 0.9, 0.38, 'Operation/\nTask')
draw_diamond(6.4, 4.1, 0.7, 0.5)
draw_rect(6.4, 3.2, 0.9, 0.38, 'Operation/\nTask')
draw_rect(6.4, 2.4, 0.9, 0.38, 'Operation/\nTask')
draw_para(8.1, 2.4, 0.8, 0.38, 'Output')

# Right Straight Arrows
draw_arrow(5.7, 5.7, 5.95, 5.7)  # Input -> Task 1
draw_arrow(6.4, 5.51, 6.4, 5.09)  # Task 1 -> Task 2
draw_arrow(6.4, 4.71, 6.4, 4.35)  # Task 2 -> Decision
draw_arrow(6.4, 3.85, 6.4, 3.39)  # Decision -> Task 3
draw_arrow(6.4, 3.01, 6.4, 2.59)  # Task 3 -> Task 4
draw_arrow(6.85, 2.4, 7.7, 2.4)  # Task 4 -> Output

# Right Decision Loopback
draw_loopback(6.75, 4.1, 7.0, 4.9, 6.85, 4.9)


# --- DOTTED EXPLOSION LINES ---
ax.plot([2.45, 4.7], [3.29, 6.4], color='black', linestyle=':', lw=1.2)
ax.plot([2.45, 4.7], [2.91, 2.1], color='black', linestyle=':', lw=1.2)

plt.tight_layout()
plt.show()
```

2- **Pareto charts** to prioritize problems or root causes. ​The Pareto chart is the source of the 80/20 rule.

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
categories = ['Hand', 'Foot', 'Eye', 'Back', 'Other']
frequencies = [300, 100, 50, 30, 20]

# Cumulative percentages
cum_percentage = np.cumsum(frequencies) / sum(frequencies) * 100

# Setup figure with warm slide background color
fig, ax1 = plt.subplots(figsize=(8, 4.5), dpi=200)
fig.patch.set_facecolor('#FFF8DC')
ax1.set_facecolor('#FFF8DC')

# Bar Plot (Frequency - Left Axis)
# width=1.0 removes gaps between bars to match the image style
bars = ax1.bar(
    categories,
    frequencies,
    color='#5B9BD5',
    edgecolor='#333333',
    width=1.0,
    align='center',
)

ax1.set_ylabel('Frequency', fontsize=9, color='#333333')
ax1.set_ylim(0, 360)
ax1.set_yticks([0, 50, 100, 150, 200, 250, 300, 350])
ax1.tick_params(axis='y', colors='#333333', labelsize=8)
ax1.tick_params(axis='x', colors='#333333', labelsize=8)
ax1.set_xlabel('Injury Type', fontsize=9, color='#333333')

# Line Plot (Cumulative Percentage - Right Axis)
ax2 = ax1.twinx()
ax2.set_facecolor('none')
ax2.plot(
    categories,
    cum_percentage,
    color='#C00000',
    marker='s',
    linewidth=1.5,
    markersize=5,
)

ax2.set_ylim(0, 105)
ax2.set_yticks(range(0, 101, 10))
ax2.set_yticklabels([f'{x}%' for x in range(0, 101, 10)])
ax2.tick_params(axis='y', colors='#333333', labelsize=8)

# Add Percentage Labels
offsets = [(12, -8), (12, -8), (10, -5), (10, -5), (-25, -2)]
for i, txt in enumerate(cum_percentage):
  ax2.annotate(
      f'{int(round(txt))}%',
      (categories[i], cum_percentage[i]),
      xytext=offsets[i],
      textcoords='offset points',
      fontsize=7.5,
      color='#555555',
  )

# Remove top spines
ax1.spines['top'].set_visible(False)
ax2.spines['top'].set_visible(False)

# Titles
fig.text(
    0.5,
    0.92,
    'Pareto Charts',
    fontsize=22,
    fontfamily='sans-serif',
    ha='center',
    va='center',
)
ax1.set_title(
    'Pareto of Injuries by Type', fontsize=11, fontweight='bold', pad=12
)

plt.tight_layout()
plt.subplots_adjust(top=0.82)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
categories = [
    'Service\nefficiency',
    'Food\nSelection',
    'Cafeteria\nenvironment',
    'Food\nquality',
]
counts = [152, 27, 20, 11]

# Calculate cumulative percentages starting from 0% at origin
total = sum(counts)
cum_percentage = [0] + list(np.cumsum(counts) / total * 100)

# Setup figure canvas with warm slide background color
fig, ax1 = plt.subplots(figsize=(8, 5.5), dpi=200)
fig.patch.set_facecolor('#FFF8DC')  # Warm yellow slide background
ax1.set_facecolor('white')

# Bar Chart (Frequency - Left Axis)
# width=1.0 eliminates gap between bars matching the exact layout
bars = ax1.bar(
    categories,
    counts,
    color='#808080',
    edgecolor='#333333',
    width=1.0,
    align='edge',
)

ax1.set_ylabel(
    'Number of ratings 1 or 2', fontsize=9.5, fontweight='bold', color='#111111'
)
ax1.set_ylim(0, 200)
ax1.set_yticks([0, 50, 100, 150, 200])
ax1.tick_params(axis='y', labelsize=8.5)

# Center Category X-Labels under each bar
ax1.set_xticks([0.5, 1.5, 2.5, 3.5])
ax1.set_xticklabels(categories, fontsize=9, fontweight='bold', color='#1A237E')

# Line Plot (Cumulative Percentage - Right Axis)
ax2 = ax1.twinx()
ax2.set_facecolor('none')

# Cumulative line points plotted across bar edges [0, 1, 2, 3, 4]
x_coords = [0, 1, 2, 3, 4]
ax2.plot(
    x_coords,
    cum_percentage,
    color='#333333',
    marker='o',
    markersize=6,
    markerfacecolor='#C00000',
    markeredgecolor='#C00000',
    linewidth=1.2,
    zorder=5,
)

ax2.set_ylim(0, 100)
ax2.set_yticks([0, 25, 50, 75, 100])
ax2.tick_params(axis='y', labelsize=8.5)

# Annotation Arrow for Cumulative percent line
ax2.annotate(
    'Cumulative\npercent line',
    xy=(1.0, cum_percentage[1]),
    xytext=(2.2, 70),
    arrowprops=dict(facecolor='black', edgecolor='black', width=1.5, headwidth=6),
    fontsize=9,
    fontweight='bold',
    color='#1A237E',
    va='center',
)

# Titles
fig.text(
    0.5,
    0.91,
    'How to use - Example',
    fontsize=24,
    fontfamily='serif',
    ha='center',
    va='center',
)
ax1.set_title(
    'Summary Results:\nStudent Satisfaction Survey',
    fontsize=10,
    fontweight='bold',
    color='#1A237E',
    pad=12,
)

plt.tight_layout()
plt.subplots_adjust(top=0.78)
plt.show()
```

​The Pareto chart begins as a bar chart, ​then the bars are arranged by ​frequency with the most frequent on the left.
 In this case, it's student dissatisfaction ​with various aspects of the cafeteria.
 ​When you've collected the counts, ​sort the data from largest to smallest.
 ​The largest bar is always on the left.
 ​You can tell at a glance that ​service efficiency is the biggest problem, ​contributing to student ​dissatisfaction with the cafeteria.
​This describes how to create a Pareto chart by hand, ​but there are software options ​that will create them automatically.

2.1 rolled throughput yield

first pass yield ​for each operation or each component. ​First pass yield is actually ​just the complement of our defects per unit calculation. ​Remember to get defects per unit, ​we divided the number of defects by the number of units.

```python
#first pass yield
  - Component 1
  - Defect/unit
  - 3/50 = 0.06 (dpu)
  - Complement = 1.0-dpu
  - 1.0-0.06 = 0.94
  - 94% good quality or first pass yield of 0.94.
```

We can get the complement of ​our DPU calculation by subtracting it from one. ​In this case, the answer is 0.94. ​We can interpret this to mean that we have ​94% good quality or first pass yield of 0.94. ​It also means that for any individual part, ​there's a 0.94 probability that it will be defect free.

```python
#first pass yield
-Compenent 1: 3/50 = 0.06 (dpu),1.0-0.06 = 0.94
-Compenent 2: 2/50 = 0.04 (dpu),1.0-0.04 = 0.96
-Compenent 3: 1/50 = 0.02 (dpu),1.0-0.02 = 0.98
-Compenent 4: 2/50 = 0.04 (dpu),1.0-0.04 = 0.96
```

The first pass yield for these components is 0.94, ​0.96, 0.98, and 0.96. ​Again, that is the first pass yield for each process. ​It's also the probability ​that a component will be defect free.

```python
#rolled throughput yield
-0.94*0.96*0.98*0.96 = 0.85
```

we can find the probability of multiple events happening ​at the same time by multiplying ​the individual probabilities. ​If we multiply the first pass yield ​or probability of good parts for each component together, ​we can get the probability that ​the assembly will be defect free. ​This is rolled throughput yield. ​The answer rounded off ​is a rolled throughput yield of 0.85. ​That is, there's only a 0.85 ​probability that these four components ​will produce a good assembly.

```python
#example

#first pass yield
-Component 1: 6/200 = 0.03 (dpu),1.0-0.03 = 0.97 (FPY)
-Component 2: 1/200 = 0.005 (dpu),1.0-0.005 = 0.995 (FPY)
-Component 3: 3/200 = 0.015 (dpu),1.0-0.015 = 0.985 (FPY)

#rolled throughput yield
- 0.97*0.995*0.985 = 0.95 (RTY)
```

There's a 0.95 probability that ​an assembly made up of ​these components will be defect free. ​Did you get it right?
