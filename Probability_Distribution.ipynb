```python
# first (graph)

The blue bars are the histogram of 40 values in this sample.

​The red line is the estimated, or fitted, population distribution. ​This represents all the coffee produced, and not just the batches in our sample. ​

location of population = Greek letter sigma
Spread = Greek letter mu
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Data matching image stats (N=40, mean=0.0832, std=0.0164)
np.random.seed(42)
data = np.random.normal(loc=0.0832, scale=0.0164, size=40)

fig, ax = plt.subplots(figsize=(7, 4))

# Histogram
counts, bins, _ = ax.hist(
    data,
    bins=8,
    density=True,
    color='#82a8d8',
    edgecolor='#555555',
    linewidth=0.8,
)

# Fitted Normal Curve
x = np.linspace(0.035, 0.125, 200)
p = norm.pdf(x, 0.0832, 0.0164)
ax.plot(x, p, color='#a61c3c', linewidth=1.5)

# Annotations: Location arrow & Spread double-arrow
ax.annotate(
    'Location',
    xy=(0.0832, 0),
    xytext=(0.0832, -4),
    arrowprops=dict(arrowstyle='->', color='#333333', lw=1.2),
    ha='center',
    va='top',
    fontsize=9.5,
    bbox=dict(boxstyle='square,pad=0.2', fc='#eeeeee', ec='none'),
)

ax.annotate(
    '',
    xy=(0.068, 15),
    xytext=(0.098, 15),
    arrowprops=dict(arrowstyle='<->', color='#555555', lw=1.5),
)
ax.text(
    0.0832,
    15,
    'Spread',
    ha='center',
    va='center',
    fontsize=9.5,
    bbox=dict(boxstyle='square,pad=0.2', fc='#eeeeee', ec='none'),
)

# Formatting
ax.set_xlim(0.035, 0.125)
ax.set_ylim(-6, 26)
ax.grid(True, color='#eaeaea', linewidth=0.8)
ax.set_axisbelow(True)

for spine in ax.spines.values():
    spine.set_color('#888888')

plt.tight_layout()
plt.show()
```

```python
#mean
based on the sample, can be used as an estimate for ​the location of the population.

#median
as an estimate for ​the unknown location of the population of produced coffee.

#Q
based on the sample we expect ​the population to have a location of 0.0832. ​Or should it be 0.0845? ​Which one is it?
```

We wish to say something about the caffeine percentage ​in the total population of coffee. ​Because we do not know what that caffeine percentage is, we select the sample of ​size 40 and measure the caffeine content for those batches. ​We analyzed the data and made a histogram and descriptive statistics.
​Next we wish to say something about the total population using this sample. ​And the location of the population is denoted by mu, and ​the spread of the population is denoted by the Greek letter sigma. ​But which descriptive statistics should we use to estimate the mu and the sigma? ​For mu, the location, we often use the mean. ​And an alternative is the median. ​So which estimator to use? ​It depends on the situation.

```python
#most often selected confidence level Is 95%.

We see that we are 95% confident that ​the actual population location will lie in ​between 0.0779 and 0.0884. ​

Thus, the mean caffeine percentage in the population will ​be in between these two values with 95 percent certainty.

```

```python
import matplotlib.pyplot as plt

# Data
y = [1, 0]
means = [0.083175, 0.0845]
lows, highs = [0.077942, 0.077411], [0.088408, 0.092]

fig, ax = plt.subplots(figsize=(7, 2))

# CI lines & points
for yi, m, l, h in zip(y, means, lows, highs):
    ax.errorbar(
        m,
        yi,
        xerr=[[m - l], [h - m]],
        fmt='o',
        color='#1f5ca9',
        capsize=4,
        ms=5,
    )

# Red arrows
kw = dict(
    arrowstyle='->', color='#a61c3c', lw=2, mutation_scale=12, zorder=4
)
ax.annotate('', xy=(0.0800, 1), xytext=(0.0779, 1), arrowprops=kw)
ax.annotate('', xy=(0.0863, 1), xytext=(0.0884, 1), arrowprops=kw)
ax.annotate('', xy=(0.0797, 0), xytext=(0.0774, 0), arrowprops=kw)
ax.annotate('', xy=(0.0895, 0), xytext=(0.0920, 0), arrowprops=kw)

# Style
ax.set_yticks(y)
ax.set_yticklabels(['Mean', 'Median'], fontweight='bold')
ax.set_xlim(0.0755, 0.0935)
ax.set_ylim(-0.5, 1.5)
ax.set_title('95% Confidence Intervals', fontweight='bold')
ax.grid(True, color='#eaeaea')
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

### M3-V3 Normal, Lognormal and Weibull distribution

Screenshot 2026-09-18 at 8.53.21 pm.png

The blue bars in the histogram show the distribution of ​the 50 values in your measured sample. ​The red curve is the population distribution that Minitab ​fitted to the data for us. ​When the number of measured values is just 50, ​you need some imagination to recognize the normal distribution. ​However, as the number of measured values becomes larger, ​the distribution converges more and more to these bell shapes. ​The more N increases, the more it starts to look like a bell shape. ​You can use a normal distribution to make some simple calculations.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import lognorm

# Generate right-skewed data
np.random.seed(42)
data = np.random.lognormal(mean=3.2, sigma=0.65, size=320)

fig, ax = plt.subplots(figsize=(6.5, 4))

# Histogram
counts, bins, _ = ax.hist(
    data,
    bins=14,
    range=(0, 150),
    color='#82a8d8',
    edgecolor='#555555',
    linewidth=0.8,
)

# Skewed Distribution Curve Fit
shape, loc, scale = lognorm.fit(data, floc=0)
x = np.linspace(0, 150, 300)
pdf = lognorm.pdf(x, shape, loc, scale) * len(data) * (bins[1] - bins[0])
ax.plot(x, pdf, color='#a61c3c', linewidth=1.5)

# Red Callout Arrow
ax.annotate(
    '',
    xy=(6, 20),
    xytext=(-25, -25),
    textcoords='offset points',
    arrowprops=dict(
        arrowstyle='->', color='#a61c3c', lw=2.5, mutation_scale=15
    ),
)

# Styling
ax.set_title('Histogram of Throughput time', fontsize=12, pad=10)
ax.set_xlabel('Throughput time', fontsize=11, fontweight='bold', labelpad=6)
ax.set_ylabel('Frequency', fontsize=11, fontweight='bold', labelpad=6)
ax.set_xticks([0, 30, 60, 90, 120, 150])
ax.set_yticks([0, 20, 40, 60, 80, 100])
ax.set_xlim(-10, 155)
ax.set_ylim(0, 105)

# Frame and Grid Setup
ax.grid(True, color='#eaeaea', linewidth=0.8)
ax.set_axisbelow(True)
for spine in ax.spines.values():
    spine.set_color('#888888')

plt.tight_layout()
plt.show()
```

```python
#lognormal distribution. ​

The Weibull distribution is often used for data, like, throughput times, and ​processing times,
       because these are often skewed variables.

# ​Many values are relatively small, and some values are very high.
```

# M3-V6 Properties of the normal distribution

Screenshot 2026-09-18 at 9.54.54 pm.png

```python
# mu = location population
# sigma = spread

the numbers were calculated directly using the Empirical Rule (68–95–99.7 Rule) shown in the top diagram.

By starting at the center mean $\mu = 0.083$ and adding/subtracting multiples of the standard deviation $\sigma = 0.016$, you get each range:

1. For 68% ($\mu \pm 1\sigma$):
 Lower: $0.083 - 0.016 = 0.067$
 Upper: $0.083 + 0.016 = 0.099$
 Range: 0.067 to 0.099

2. For 95% ($\mu \pm 2\sigma$):
 Lower: $0.083 - 2(0.016) = 0.083 - 0.032 = 0.051$
 Upper: $0.083 + 2(0.016) = 0.083 + 0.032 = 0.115$
 Range: 0.051 to 0.115

3. For 99% ($\mu \pm 2.5\sigma$):
 Lower: $0.083 - 2.5(0.016) = 0.083 - 0.040 = 0.043$
 Upper: $0.083 + 2.5(0.016) = 0.083 + 0.040 = 0.123$
 Range: 0.043 to 0.123

4. For 99.7% ($\mu \pm 3\sigma$):
 Lower: $0.083 - 3(0.016) = 0.083 - 0.048 = 0.035$
 Upper: $0.083 + 3(0.016) = 0.083 + 0.048 = 0.131$
 Range: 0.035 to 0.131
```
