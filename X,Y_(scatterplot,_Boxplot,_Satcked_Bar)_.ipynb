```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data generation matching trend
np.random.seed(42)
x = np.random.uniform(25, 45, 40)
y = 0.20 - 0.0035 * x + np.random.normal(0, 0.005, 40)

fig, ax = plt.subplots(figsize=(6, 4))

# Scatter plot
ax.scatter(x, y, color='#1F5CA9', s=25)

# Red arrows
ax.annotate('', xy=(43.5, 0.042), xytext=(26.5, 0.11),
            arrowprops=dict(arrowstyle='->', color='#A6192E', lw=3))
ax.annotate('', xy=(43.5, 0.03), xytext=(25, 0.03),
            arrowprops=dict(arrowstyle='->', color='#A6192E', lw=3))

# Labels & Limits
ax.set_xlabel('Extraction time', fontsize=12)
ax.set_ylabel('Caffeine Content', fontsize=12)
ax.set_xlim(24, 46)
ax.set_ylim(0.028, 0.122)

# Spines setup (no top/right borders)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

​the caffeine content decreases as the extraction time increases.

Y (numeric) , X (categoric) = Boxplot

The caffeine content in the coffee is still our Y variable, CTQ. ​However, now I'm studying a different influence factor. ​I study the extractor number, the machine number which the coffee is put into. ​This is a categorical influence factor or X variable.

```python
import matplotlib.pyplot as plt

# Data
data = [
    [0.026, 0.041, 0.051, 0.055, 0.065],
    [0.046, 0.059, 0.069, 0.075, 0.095],
    [0.112, 0.122, 0.131, 0.139, 0.147],
]

# Plot setup
fig, ax = plt.subplots(figsize=(6, 4))
bp = ax.boxplot(
    data,
    patch_artist=True,
    widths=0.3,
    boxprops=dict(facecolor="#8da9c4", color="#555555"),
    medianprops=dict(color="#2b4c7e", linewidth=1.5),
    whiskerprops=dict(color="#555555"),
    capprops=dict(color="#555555"),
)

# Reference line
ax.axhline(y=0.1, color="#a61c3c", linewidth=2.5)

# Styling
ax.set_title("Boxplot of Caffeine%", fontsize=13)
ax.set_xlabel("Extractor nr.", fontsize=11)
ax.set_ylabel("Caffeine%", fontsize=11)
ax.set_xticklabels([1, 2, 3])
ax.set_ylim(0.018, 0.162)

plt.tight_layout()
plt.show()
```

The red line = caffeine content to always be below 0.1%

### Y (categoric)  X ( Numeric) = Boxplot (Group) or Transpot Boxplot

Q = We would like to know if the student's math grade in high school ​affects the likelihood that they pass the first year at university.

```python
import matplotlib.pyplot as plt

# Data estimated from boxplot
data_no = [5.0, 6.0, 7.0, 8.0, 9.0]
data_yes = [7.0, 8.0, 9.0, 10.0, 10.0]

fig, ax = plt.subplots(figsize=(6.5, 4))

# Horizontal boxplot
bp = ax.boxplot(
    [data_yes, data_no],
    vert=False,
    patch_artist=True,
    widths=0.35,
    boxprops=dict(facecolor="#79a3d4", edgecolor="#444444", linewidth=0.8),
    medianprops=dict(color="#333333", linewidth=1.2),
    whiskerprops=dict(color="#444444", linewidth=0.8),
    capprops=dict(color="#444444", linewidth=0.8),
)

# Formatting
ax.set_yticklabels(["Yes", "No"], fontsize=10, fontweight="bold", color="#333333")
ax.set_xlabel("Math grade", fontsize=11, fontweight="bold", color="#444444", labelpad=8)
ax.set_ylabel("Pass?", fontsize=11, fontweight="bold", color="#444444", labelpad=8)
ax.set_xlim(4.75, 10.3)
ax.set_xticks([5, 6, 7, 8, 9, 10])

# Light grid and crisp frame
ax.grid(True, color="#eaeaea", linewidth=0.8)
ax.set_axisbelow(True)
for spine in ax.spines.values():
    spine.set_color("#888888")
    spine.set_linewidth(1)

plt.tight_layout()
plt.show()
```

This is our transposed Boxplot which means that we have our passing in our vertical ​axis and a math grade on our horizontal axis.

We see that the students who pass the first year ​have higher grades than students who didn't pass the first year.

### X,Y (categorical) = stacked bar chart

```python
# Patient number,

# the Specialist who treated the patient

# Department that the patient was lying.


we're interested to know if each specialist ​goes to the different departments equally frequently. ​
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Data
df = pd.DataFrame({
    'Specialist': ['P', 'Q', 'R'],
    'A1': [81, 52, 78],
    'A4': [17, 15, 26],
})

fig, ax = plt.subplots(figsize=(6, 4))

# Stacked Bar Plot
ax.bar(
    df['Specialist'],
    df['A1'],
    width=0.4,
    color='#79a3d4',
    edgecolor='#444444',
    linewidth=0.8,
    label='A1',
)
ax.bar(
    df['Specialist'],
    df['A4'],
    width=0.4,
    bottom=df['A1'],
    color='#c84b4b',
    edgecolor='#444444',
    linewidth=0.8,
    label='A4',
)

# Styling
ax.set_title('Chart of Specialist, Department', fontsize=12, pad=10)
ax.set_xlabel(
    'Specialist',
    fontsize=10,
    fontweight='bold',
    color='#444444',
    loc='left',
    labelpad=8,
)
ax.set_ylabel('Count', fontsize=10, fontweight='bold', color='#444444', labelpad=8)
ax.set_ylim(0, 110)

# Legend (matching order in image: A4 top, A1 bottom)
handles, labels = ax.get_legend_handles_labels()
ax.legend(
    handles[::-1],
    labels[::-1],
    title='Department',
    loc='upper left',
    bbox_to_anchor=(1, 1),
    frameon=False,
)

# Grid & Spines
ax.grid(True, color='#eaeaea', linewidth=0.8)
ax.set_axisbelow(True)
for spine in ax.spines.values():
    spine.set_color('#888888')
    spine.set_linewidth(1)

plt.tight_layout()
plt.show()
```
