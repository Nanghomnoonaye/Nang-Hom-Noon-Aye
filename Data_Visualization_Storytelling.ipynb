# Misleading

```python
years = [2023,2024]
revenue = [100, 110]

plt.figure(figsize=(6,4))
plt.ylim(95, 112)
plt.bar(years, revenue)

plt.title('Misleading Revenue')
plt.xlabel('Year')
plt.ylabel('Revenue')

plt.show()
```

```python
# Data: years and revenue
years = ['2023','2024']
revenue = [100, 110]

# Create bar chart with y-axis starting at 95 instead of 0
plt.bar(years, revenue, color='orange')
plt.ylim(95, 115)  # Y-axis limits start close to the data, exaggerating difference

plt.title('Misleading Revenue')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.show()
```

# Correct revenue

```python
plt.figure(figsize=(6,4))
plt.ylim(0, 120)  # Start y-axis at zero for honest representation
plt.bar(years, revenue)

plt.title('Correct Revenue Chart')
plt.xlabel('Year')
plt.ylabel('Revenue')

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
south_sales = [110, 115, 100, 90, 80]  # Example sales data for South Region
north_sales = [100, 120, 140, 160, 180] # Example sales data for North Region

plt.figure(figsize=(7,4))
plt.plot(months,
         south_sales,
         marker='o',
         label='South Region')
plt.plot(months,
         north_sales,
         marker='o',
         label='North Region')

plt.title('Regional Sales')
plt.ylabel('Sales')
plt.xlabel('Month')
plt.legend()
plt.show()
```

# Context -> Conflict -> Resolution

```python
days=['Mon','Tue','Wed','Thu','Fri','Sat','Sun']
traffic = [1000,1100,1200,1300,1400,1600,1700]
sales=[100,120,130,140,150,90,80]

fig, ax1 = plt.subplots(figsize=(8,4))

ax1.plot(days,
         traffic,
         marker='o',
         label='Traffic')

ax1.set_ylabel('Traffic')

ax2 = ax1.twinx()

ax2.plot(days,
         sales,
         marker='o',
         label='Sales')

ax2.set_ylabel('Sales')

plt.title('Website Traffic vs Sales')
plt.show()
```

### Story Line * situation -> problem -> insight ->action

### Let's take ​an example

1.   Website traffic increased.this is the **situation**.
2.   **Problem**, sales did not increase.


1.   **Insight**, most visitor leaves before checkout.
2.   **​Action**, improve the checkout process. This is complete data story.
